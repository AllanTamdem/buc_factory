```python
import openai

client = openai.OpenAI()
models = sorted(m.id for m in client.models.list())
print("All models:")
for m in models:
    print(m)
chat_models = sorted(m.id for m in client.models.list() if "gpt" in m.id or m.id.startswith("o"))
print("\nChat models:")
for m in chat_models:
    print(m)
```
